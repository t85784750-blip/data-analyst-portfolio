```python
name = input("Enter your name: ")

weight = int(input ("Enter your weight in kilograms: "))

height = int(input("Enter your height in centimetres: "))

BMI = (weight) / (height / 100 * height / 100) 

print(BMI)

if BMI > 0:
    if BMI < 16:
        print(f"{name}, you have Severe Thinness. Mate, you're basically a walking skeleton—go hug a burger (or three)!")
    elif BMI <= 17:
        print(f"{name}, you have Moderate Thinness. A stiff breeze could carry you away. Time to add some snacks to your life!")
    elif BMI <= 18.5:
        print(f"{name}, you have Mild Thinness. You're skating on the edge of 'needs more pizza.' Consider it a friendly nudge.")
    elif BMI <= 25:
        print(f"{name}, you are normal weight. Gold star! You're in the sweet spot—keep doing whatever magic you're doing.")
    elif BMI <= 30:
        print(f"{name}, you are overweight. Those extra pounds are just love handles looking for attention. Ditch the late-night ice cream and go for a jog—your couch will miss you, but your jeans won't!")
    elif BMI <= 35:
        print(f"{name}, you are obese class I. You've got some extra cushion for the pushin', but maybe swap the donuts for dumbbells? Your future self will thank you.")
    elif BMI <= 40:
        print(f"{name}, you are obese class II. You're carrying a bit more than the recommended luggage. Time to lighten the load—think salads, squats, and saying no to second helpings!")
    else:
        print(f"{name}, you are obese class III. Whoa there, champ! You've officially unlocked VIP status in the 'extra fluffy' club. Let's start moving more and munching less— you've got this!")
```
